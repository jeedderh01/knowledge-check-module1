# 1. Five limitations of Excel that make Python essential for predictive AI in offshore engineering


1. Excel cannot handle a very large datasets:	A SCADA system monitoring five subsea-manifold sensors continuously can generate millions of data points over time. Managing and processing such datasets becomes difficult in Excel.
2. It has limited automation and scalability:	An offshore engineer may need to repeatedly calculate pipeline properties, temperatures, pressures, or equipment-performance indicators for thousands of records. Python can automate these calculations with functions and loops.
3. It has limited machine-learning capabilities:	Predictive maintenance may require machine-learning models to predict equipment failure from pressure, temperature, vibration, and flow data. Python provides libraries and frameworks for developing these models.
4. It has a greater risk of manual errors in complex calculations:	A large Excel workbook containing formulas for pipeline design, wall thickness, and operating conditions can be difficult to audit. A Python function can define a calculation once and apply it consistently.
5. It has a limited integration with engineering and AI workflows:	Predictive AI may require data cleaning, statistical analysis, visualization, machine learning, and integration with SCADA or other data sources. Python can connect these stages into an automated workflow.

## 2. Fahrenheit-to-Celsius Python Function

The formula for converting Fahrenheit to Celsius is (temperature - 32) * 5 / 9


```python
def fahrenheit_to_celsius(temperature: float) -> float:
    """Convert temperature from Fahrenheit to Celsius.

    Args:
        temperature: Temperature in degrees Fahrenheit.

    Returns:
        Temperature in degrees Celsius.

    Raises:
        TypeError: If the input is not a number.
    """
    if not isinstance(temperature, (int, float)):
        raise TypeError("Temperature must be a number.")

    return (temperature - 32) * 5 / 9


# Test 1: Valid input
print(fahrenheit_to_celsius(212))

# Test 2: Valid input
print(fahrenheit_to_celsius(150))

# Test 3: Invalid input
try:
    print(fahrenheit_to_celsius("re"))
except TypeError as error:
    print(f"Error: {error}")
```

# 7. Why version control is essential for engineering software

Version control is important because engineering software can contain calculations that influence design, operation, maintenance, and safety decisions. Git records changes and makes it possible to identify, compare, and restore previous versions.

#Scenario 1: Incorrect engineering formula

Suppose an engineer changes a pipeline wall-thickness formula and accidentally introduces an error.

With Git, the team can:

Identify who made the change.
See exactly what changed.
Compare the new calculation with the previous version.
Restore the previous working version.

This can prevent an incorrect calculation from remaining in the production code

#Scenario 2: Two engineers modify the same predictive-maintenance program

Two engineers may simultaneously modify a Python program that predicts subsea equipment failure.

Without version control, one engineer could accidentally overwrite the other's work.

Git allows the changes to be tracked and merged, while conflicts can be identified and resolved.

#Scenario 3: A new AI model produces incorrect predictions

An engineering team may replace an existing predictive-maintenance model with a new machine-learning model.

If the new model performs incorrectly, Git allows the team to return to the previously validated version while investigating the problem.

# 9. The try/except pattern

The try/except pattern allows a Python program to handle errors without necessarily stopping the entire program.

The code that might produce an error is placed inside the try block. If an expected error occurs, the except block handles it.

```python
def safe_divide(number1: float, number2: float) -> float | str:
    """Safely divide two numbers.

    Args:
        number1: The numerator.
        number2: The denominator.

    Returns:
        The result of the division or an error message.
    """
    try:
        return number1 / number2
    except ZeroDivisionError:
        return "Error: Division by zero"


# Test the function
print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

# 10. Tufte's data-ink ratio principle

Tufte’s **data-ink ratio** principle states that a good chart should use most of its visual elements to communicate the actual data. Unnecessary decoration, such as excessive colours, 3D effects, heavy gridlines, and distracting backgrounds, should be minimized.

For a chart showing the **distribution of significant wave heights in the West African basin**, I would apply the principle by:

* Using a simple **histogram** to show how frequently different wave heights occur.
* Clearly labelling the x-axis as **Significant Wave Height (m)**.
* Clearly labelling the y-axis as **Frequency** or **Number of Observations**.
* Using appropriate wave-height intervals (bins) so the distribution is easy to interpret.
* Removing unnecessary 3D effects, decorative images, and excessive gridlines.
* Using a clear title such as **“Distribution of Significant Wave Heights in the West African Basin.”**
* Keeping the wave-height data as the main visual focus.

This approach would allow an offshore engineer to quickly identify the **most common wave-height ranges, the spread of the data, and unusually high wave-height observations** without being distracted by unnecessary visual elements.
